# ⚙️ Feature Engineering
Building 13 new features on top of the raw dataset.

```
from src.data.loader import load_raw_data
from src.data.preprocessor import clean_data, get_features_and_target
from src.features.engineering import engineer_features
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = clean_data(load_raw_data())
df_eng = engineer_features(df)
print(f"Before: {df.shape[1]} features")
print(f"After:  {df_eng.shape[1]} features")
print(f"Added:  {df_eng.shape[1] - df.shape[1]} new features")
```

```
# New features summary
new_features = ['tenure_cohort','avg_monthly_revenue','revenue_per_product',
    'charge_increase_rate','product_count','risk_score',
    'is_loyal','is_high_spender','is_senior_alone',
    'is_month_to_month','is_electronic_check','is_fiber']
df_eng[new_features].describe().round(2)
```

```
# Risk score distribution by churn
plt.figure(figsize=(10, 5))
sns.boxplot(data=df_eng, x='Churn', y='risk_score', palette=['#2ecc71','#e74c3c'])
plt.title('Risk Score Distribution by Churn')
plt.tight_layout()
plt.show()
```

```
# Product count vs churn
plt.figure(figsize=(10, 5))
sns.countplot(data=df_eng, x='product_count', hue='Churn', palette=['#2ecc71','#e74c3c'])
plt.title('Product Count vs Churn')
plt.tight_layout()
plt.show()
```

```
# Tenure cohort vs churn rate
churn_by_cohort = df_eng.groupby('tenure_cohort')['Churn'].mean().reset_index()
plt.figure(figsize=(8, 4))
plt.bar(churn_by_cohort['tenure_cohort'], churn_by_cohort['Churn'], color='#e74c3c')
plt.xlabel('Tenure Cohort (1=0-12m, 2=12-24m, 3=24-48m, 4=48-72m)')
plt.ylabel('Churn Rate')
plt.title('Churn Rate by Tenure Cohort')
plt.tight_layout()
plt.show()
```
