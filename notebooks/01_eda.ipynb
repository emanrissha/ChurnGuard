# 📊 Exploratory Data Analysis
Understanding the Telco Customer Churn dataset before modelling.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.data.loader import load_raw_data

sns.set_theme(style="whitegrid")
df = load_raw_data()
print(df.shape)
df.head()
```

```
# Basic info
print(df.dtypes)
print('\nMissing values:')
print(df.isnull().sum())
```

```
# Churn distribution
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
df['Churn'].value_counts().plot(kind='bar', ax=ax[0], color=['#2ecc71','#e74c3c'])
ax[0].set_title('Churn Count')
df['Churn'].value_counts(normalize=True).plot(kind='pie', ax=ax[1], autopct='%1.1f%%', colors=['#2ecc71','#e74c3c'])
ax[1].set_title('Churn Rate')
plt.tight_layout()
plt.show()
```

```
# Numerical features distribution
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
df[num_cols] = df[num_cols].apply(pd.to_numeric, errors='coerce')
df[num_cols].hist(bins=30, figsize=(14, 4), color='#3498db', edgecolor='white')
plt.suptitle('Numerical Features Distribution')
plt.tight_layout()
plt.show()
```

```
# Churn by contract type
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Contract', hue='Churn', palette=['#2ecc71','#e74c3c'])
plt.title('Churn by Contract Type')
plt.tight_layout()
plt.show()
```

```
# Churn by tenure
plt.figure(figsize=(12, 5))
sns.boxplot(data=df, x='Churn', y='tenure', palette=['#2ecc71','#e74c3c'])
plt.title('Tenure Distribution by Churn')
plt.tight_layout()
plt.show()
```

```
# Churn by internet service
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='InternetService', hue='Churn', palette=['#2ecc71','#e74c3c'])
plt.title('Churn by Internet Service')
plt.tight_layout()
plt.show()
```

```
# Monthly charges vs tenure
plt.figure(figsize=(10, 6))
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
scatter = plt.scatter(df['tenure'], df['MonthlyCharges'], c=df['Churn'].map({'Yes':1,'No':0}), cmap='RdYlGn_r', alpha=0.5)
plt.colorbar(scatter, label='Churn (1=Yes)')
plt.xlabel('Tenure (months)')
plt.ylabel('Monthly Charges (₪)')
plt.title('Tenure vs Monthly Charges — colored by Churn')
plt.tight_layout()
plt.show()
```

```
# Key insights
print("=== KEY EDA INSIGHTS ===")
print(f"Total customers: {len(df):,}")
print(f"Churn rate: {(df['Churn']=='Yes').mean():.1%}")
print(f"Avg tenure (churned): {df[df['Churn']=='Yes']['tenure'].mean():.1f} months")
print(f"Avg tenure (retained): {df[df['Churn']=='No']['tenure'].mean():.1f} months")
print(f"Month-to-month churn rate: {(df[df['Contract']=='Month-to-month']['Churn']=='Yes').mean():.1%}")
```
